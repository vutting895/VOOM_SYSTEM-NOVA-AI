```python

# Loop through each row in the dataframe
for i in range(3, len(df)):
    # Check if a bullish or bearish harmonic pattern exists
    if harmonicpatterns.is_bullish_harmonic_pattern(df.iloc[i-3:i]):
        signals.append((df.index[i], 'Buy'))
    elif harmonicpatterns.is_bearish_harmonic_pattern(df.iloc[i-3:i+1, :]):
        signals.append((df.index[i], 'Sell'))
    else:
        signals.append((df.index[i], 'None'))
```

```python
# Create a list to store the signals
signals = []

# Create a DataFrame of signals
signals_df = pd.DataFrame(signals, columns=['bullish_harmonic_pattern', bearish_harmonic_pattern'])
signals_df.set_index('Entry Time', inplace=True)

# Create a DataFrame of prices
prices_df = pd.DataFrame({'h': df['Close']}, index=df.index)

# Combine signals and prices DataFrames
df = pd.concat([signals_df, prices_df], axis=1)

# Forward fill the signals
df['Signal'].fillna(method='ffill', inplace=True)

# Backtest the signals
df['Returns'] = np.log(df['Close'] / df['Close'].shift(1))

# Calculate strategy returns
df['Strategy Returns'] = df['Returns'] * np.where(df['Signal'] == 'Buy', 1, -1)

# Calculate cumulative returns
df['Cumulative Returns'] = df['Strategy Returns'].cumsum()

# Plot cumulative returns
df['Cumulative Returns'].plot(figsize=(10, 6), title='Harmonic Pattern Trading Strategy')

# Show the plot
plt.show()
```
